```python
import pandas as pd
df1 = pd.DataFrame({"Name": ["Amit", "Priya", "Rahul"], "Sales": [1000, 1500, 2000]})
df2 = pd.DataFrame({"Name": ["Sneha", "Arjun", "Neha"], "Sales": [1800, 2200, 1700]})
print("DataFrame 1:")
print(df1)
print("DataFrame 2:")
print(df2)
```

```python
vertical_concat = pd.concat([df1, df2], axis=0)
print(vertical_concat)
```

```python
vertical_reset = pd.concat([df1, df2], axis=0, ignore_index=True)
print(vertical_reset)
```

```python
horizontal_concat = pd.concat([df1, df2], axis=1)
print(horizontal_concat)
```

```python
key_concat = pd.concat([df1, df2], keys=["Region_A", "Region_B"])
print(key_concat)
```

```python
df3 = pd.DataFrame({"Name": ["Kiran", "Divya"], "Department": ["IT", "HR"]})
different_columns = pd.concat([df1, df3], ignore_index=True)
print(different_columns)
```

```python
inner_concat = pd.concat([df1, df3], join="inner", ignore_index=True)
print(inner_concat)
```

```python
outer_concat = pd.concat([df1, df3], join="outer", ignore_index=True)
print(outer_concat)
```

```python
df4 = pd.DataFrame({"Name": ["Riya", "Vijay"], "Sales": [2500, 1900]})
multiple_concat = pd.concat([df1, df2, df4], ignore_index=True)
print(multiple_concat)
```

```python
sorted_concat = pd.concat([df1, df3], ignore_index=True, sort=True)
print(sorted_concat)
```

```python
selected_concat = pd.concat([df1[["Name"]], df2[["Name"]]], ignore_index=True)
print(selected_concat)
```

```python
print("DataFrame 1:")
print(df1)
print("DataFrame 2:")
print(df2)
print("Final Concatenated DataFrame:")
print(pd.concat([df1, df2], ignore_index=True))
```
